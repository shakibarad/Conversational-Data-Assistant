```python
tables = {
    "orders": order,
    "customers": customer,
    "order_items": order_items,
    "products": product,
    "sellers": seller,
    "payments": payment,
    "reviews": review,
    "geolocation": geolocation,
    "category_translation": category_translation
}

for name, df in tables.items():
    print(f"{name}: {df.shape}")
```

```python
for table1_name, table1 in tables.items():
    
    for table2_name, table2 in tables.items():

        if table1_name >= table2_name:
            continue

        common_columns = list(
            set(table1.columns) & set(table2.columns)
        )

        if common_columns:
            print(f"{table1_name} ↔ {table2_name}")
            print("Common columns:", common_columns)
            print("-" * 50)
```

```python
relationships = []

for table1_name, table1 in tables.items():

    for table2_name, table2 in tables.items():

        if table1_name >= table2_name:
            continue

        common_columns = list(
            set(table1.columns) & set(table2.columns)
        )

        if common_columns:
            relationships.append({
                "table_1": table1_name,
                "table_2": table2_name,
                "common_columns": common_columns
            })

relationships_df = pd.DataFrame(relationships)

relationships_df
```

```python
relationships = [
    ("customers", "customer_id", "orders", "customer_id"),
    ("orders", "order_id", "order_items", "order_id"),
    ("orders", "order_id", "payments", "order_id"),
    ("orders", "order_id", "reviews", "order_id"),
    ("products", "product_id", "order_items", "product_id"),
    ("sellers", "seller_id", "order_items", "seller_id"),
    ("products", "product_category_name", "category_translation", "product_category_name")
]
```

```python
for parent_table, parent_key, child_table, child_key in relationships:
    
    parent = tables[parent_table]
    child = tables[child_table]

    parent_unique = parent[parent_key].nunique()
    child_unique = child[child_key].nunique()

    unmatched = ~child[child_key].isin(
        parent[parent_key].dropna()
    )

    print(f"{parent_table} → {child_table}")
    print(f"Key: {parent_key}")
    print(f"Parent unique values: {parent_unique}")
    print(f"Child unique values: {child_unique}")
    print(f"Unmatched child keys: {unmatched.sum()}")
    print("-" * 60)
```

```python
from graphviz import Digraph

schema = Digraph("Olist Schema")

for table_name in tables:
    schema.node(table_name, table_name)

for parent_table, parent_key, child_table, child_key in relationships:
    schema.edge(
        parent_table,
        child_table,
        label=f"{parent_key} → {child_key}"
    )

schema
```

## 1. What percentage of orders are delivered late?

```python
order["order_delivered_customer_date"] = pd.to_datetime(
    order["order_delivered_customer_date"]
)

order["order_estimated_delivery_date"] = pd.to_datetime(
    order["order_estimated_delivery_date"]
)

late_orders = (
    order["order_delivered_customer_date"]
    > order["order_estimated_delivery_date"]
).sum()

total_delivered_orders = order["order_delivered_customer_date"].notna().sum()

late_delivery_percentage = (
    late_orders / total_delivered_orders
) * 100

print("Late orders:", late_orders)
print("Total delivered orders:", total_delivered_orders)
print("Late delivery percentage:", round(late_delivery_percentage, 2), "%")
```

## 2. How many days late are they?

```python
order["delay_days"] = (
    order["order_delivered_customer_date"]
    - order["order_estimated_delivery_date"]
).dt.days

order[["order_delivered_customer_date",
       "order_estimated_delivery_date",
       "delay_days"]].head()
```

## 3. Which regions have the most delays?

```python
customer = pd.read_csv (r"C:\Users\shate\Desktop\wgls-ai-logistics-assistant\data\raw\olist_customers_dataset.csv")
```

```python
order_customer = order.merge(
    customer,
    on="customer_id",
    how="left"
)

order_customer.head()
```

```python
state_performance = (
    order_customer
    .groupby("customer_state")
    .agg(
        total_orders=("order_id", "count"),
        late_orders=("delay_days", lambda x: (x > 0).sum())
    )
)

state_performance["late_delivery_rate"] = (
    state_performance["late_orders"]
    / state_performance["total_orders"]
    * 100
)

state_performance = state_performance.sort_values(
    "late_delivery_rate",
    ascending=False
)

state_performance.head(10)
```

## 4. Are delays getting worse over time?

```python
order["order_purchase_timestamp"] = pd.to_datetime(
    order["order_purchase_timestamp"]
)

order["order_month"] = (
    order["order_purchase_timestamp"]
    .dt.to_period("M")
    .astype(str)
)

order[["order_purchase_timestamp", "order_month"]].head()
```

```python
# Calculate monthly late-delivery rate for each state
monthly_state = (
    order_customer
    .groupby(["customer_state", "order_month"])
    .agg(
        total_orders=("order_id", "count"),
        late_orders=("delay_days", lambda x: (x > 0).sum())
    )
    .reset_index()
)

monthly_state["late_rate"] = (
    monthly_state["late_orders"]
    / monthly_state["total_orders"]
    * 100
)

# Sort by state and month
monthly_state = monthly_state.sort_values(
    ["customer_state", "order_month"]
)

monthly_state.head(20)
```

```python
# Compare first and last month for each state
state_trends = (
    monthly_state
    .groupby("customer_state")
    .agg(
        first_month_rate=("late_rate", "first"),
        last_month_rate=("late_rate", "last")
    )
)

# Calculate the change
state_trends["change"] = (
    state_trends["last_month_rate"]
    - state_trends["first_month_rate"]
)

# Keep states where the late-delivery rate increased
states_getting_worse = state_trends[
    state_trends["change"] > 0
].index.tolist()

print("States where delays increased:")
print(states_getting_worse)
```

```python
last_3_months = (
    monthly_state
    .groupby("customer_state")
    .tail(3)
)

last_3_months.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Create table: states as rows, months as columns
heatmap_data = last_3_months.pivot(
    index="customer_state",
    columns="order_month",
    values="late_rate"
)

# Plot heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    heatmap_data,
    annot=True,
    fmt=".1f",
    cmap="Reds"
)

plt.title("Late Delivery Rate by State - Last 3 Months")
plt.xlabel("Month")
plt.ylabel("State")

plt.show()
```

## 5. Are certain sellers associated with delays?

```python
seller_orders = (
    order
    .merge(
        order_items[["order_id", "seller_id"]],
        on="order_id",
        how="inner"
    )
    .merge(
        seller,
        on="seller_id",
        how="left"
    )
)

seller_orders.head()
```

```python
seller_delay_data = (
    order
    .merge(
        customer[["customer_id", "customer_state"]],
        on="customer_id",
        how="left"
    )
    .merge(
        order_items[["order_id", "seller_id"]],
        on="order_id",
        how="inner"
    )
    .merge(
        seller[["seller_id", "seller_city", "seller_state"]],
        on="seller_id",
        how="left"
    )
)
```

```python
seller_delay_data = seller_delay_data[
    [
        "order_id",
        "seller_id",
        "seller_city",
        "seller_state",
        "customer_state",
        "order_delivered_customer_date",
        "order_estimated_delivery_date",
        "delay_days"
    ]
].copy()

seller_delay_data.head()
```

```python
same_state_delays = seller_delay_data[
    (seller_delay_data["seller_state"] == seller_delay_data["customer_state"])
    & (seller_delay_data["delay_days"] > 0)
].copy()
```

```python
same_state_delays[
    [
        "order_id",
        "seller_id",
        "seller_city",
        "seller_state",
        "customer_state",
        "delay_days"
    ]
].sort_values(
    "delay_days",
    ascending=False
).head(20)
```

```python
different_state_delays = seller_delay_data[
    (seller_delay_data["seller_state"] != seller_delay_data["customer_state"])
    & (seller_delay_data["delay_days"] > 0)
].copy()
```

```python
different_state_delays[
    [
        "order_id",
        "seller_id",
        "seller_city",
        "seller_state",
        "customer_state",
        "delay_days"
    ]
].sort_values(
    "delay_days",
    ascending=False
).head(20)
```

```python
delay_group_summary = (
    seller_delay_data[
        seller_delay_data["delay_days"] > 0
    ]
    .assign(
        route_type=lambda x: x.apply(
            lambda row: "Same State"
            if row["seller_state"] == row["customer_state"]
            else "Different State",
            axis=1
        )
    )
    .groupby("route_type")
    .agg(
        delayed_orders=("order_id", "nunique"),
        average_delay_days=("delay_days", "mean")
    )
    .reset_index()
)

delay_group_summary
```

```python
seller_performance = (
    seller_delay_data[
        seller_delay_data["delay_days"].notna()
    ]
    .groupby("seller_id")
    .agg(
        delivered_orders=("order_id", "nunique"),
        late_orders=("delay_days", lambda x: (x > 0).sum()),
        average_delay_days=("delay_days", "mean")
    )
    .reset_index()
)

seller_performance["late_delivery_rate"] = (
    seller_performance["late_orders"]
    / seller_performance["delivered_orders"]
    * 100
)
```

```python
seller_performance_reliable = seller_performance[
    seller_performance["delivered_orders"] >= 20
].copy()
```

```python
seller_performance_reliable.sort_values(
    "late_delivery_rate",
    ascending=False
).head(20)
```
